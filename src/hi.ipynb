```python
import mne
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 1. READ EDF FILE
# =========================

file_path = "/Users/audaodoantruc/Downloads/IU-MIBCI_4class/G1/F11_A.edf"

raw = mne.io.read_raw_edf(file_path, preload=True)

print("Sampling rate:", raw.info["sfreq"])
print("Total channels:", len(raw.ch_names))


# =========================
# 2. SELECT EEG CHANNELS
# =========================

eeg_channels = [
'Cz','Fz','Fp1','F7','F3','FC1','C3','FC5','FT9','T7',
'CP5','CP1','P3','P7','PO9','O1','Pz','Oz','O2','PO10',
'P8','P4','CP2','CP6','T8','FT10','FC6','C4','FC2',
'F4','F8','Fp2']
raw_eeg = raw.copy().pick_channels(eeg_channels)

print("EEG channels:", raw_eeg.ch_names)


# =========================
# 3. BANDPASS FILTER
# =========================

raw_filt = raw_eeg.copy().filter(
    l_freq=8,
    h_freq=30
)


# =========================
# 4. NOTCH FILTER (POWERLINE)
# =========================

raw_filt.notch_filter(freqs=50)


# =========================
# 5. ICA ARTIFACT REMOVAL
# =========================

ica = mne.preprocessing.ICA(
    n_components=20,
    random_state=97,
    max_iter="auto"
)

ica.fit(raw_filt)

# detect artifact automatically
# The following lines are commented out because no EOG channels were found
# in the data after picking only EEG channels.
# If your original EDF file contains EOG channels, you should include them
# in the `eeg_channels` list to enable automatic EOG artifact detection.
# Alternatively, you can manually inspect ICA components and exclude artifacts.
# eog_indices, eog_scores = ica.find_bads_eog(raw_filt)

# print("Detected artifact components:", eog_indices)

# ica.exclude = eog_indices


# =========================
# 6. APPLY ICA
# =========================

raw_clean = raw_filt.copy()

ica.apply(raw_clean)


# =========================
# 7. GET CLEAN DATA
# =========================

data1 = raw_clean.get_data()

print("Clean EEG shape:", data1.shape)


# =========================
# 8. PLOT EEG
# =========================

sfreq = raw_clean.info["sfreq"]

time = np.arange(data1.shape[1]) / sfreq

plt.figure(figsize=(14,10))

for i in range(len(raw_clean.ch_names)):

    plt.subplot(len(raw_clean.ch_names),1,i+1)

    plt.plot(time, data1[i])

    plt.ylabel(raw_clean.ch_names[i], rotation=0, labelpad=25)

plt.xlabel("Time (s)")

plt.tight_layout()

plt.show()
```

```python
import numpy as np
import mne

# Filter for 'MI' type markers
mi_markers_df = marker_df[marker_df['type'] == 'MI'].copy()

# Define event IDs based on marker_value
# 0: Tay phải, 1: Tay trái, 2: Chân phải, 3: Chân trái
event_id = {
    'Tay phải': 0,
    'Tay trái': 1,
    'Chân phải': 2,
    'Chân trái': 3
}

# Get sampling frequency from the cleaned raw data
sfreq = raw_clean.info['sfreq']

events = []
for index, row in mi_markers_df.iterrows():
    onset_s = row['latency']
    marker_value = row['marker_value']

    # Convert onset from seconds to samples
    onset_sample = int(onset_s * sfreq)

    # Append event in MNE format: [onset_sample, duration_sample (0), event_id]
    # Using marker_value directly as the event_id in the array
    events.append([onset_sample, 0, marker_value])

events = np.array(events)

# Define the time window for epochs relative to the event onset
# Example: -0.5s to 3.0s around the event
tmin = -0.5
tmax = 3.0

# Create epochs
# The `event_id` dictionary maps the integer values in `events` to descriptive labels
epochs = mne.Epochs(raw_clean, events, event_id=event_id, tmin=tmin, tmax=tmax,
                    preload=True, baseline=(None, 0), verbose=False)

print("Number of epochs extracted:", len(epochs))
print("Epochs info:")
print(epochs)

# Optional: Plot evoked responses for each condition to verify
# This will generate separate plots for each condition
for condition in event_id.keys():
    # Correctly check if there are epochs for the current condition label
    epochs_for_condition = epochs[condition]
    if len(epochs_for_condition) > 0:
        epochs_for_condition.average().plot(picks='eeg', titles=f'Evoked for {condition}')
    else:
        print(f"No epochs found for condition: {condition}")
```

```python
import numpy as np
import pandas as pd
from scipy.linalg import solve
import matplotlib.pyplot as plt

# ===========================================================
# 1️⃣ DANH SÁCH LABEL (classes)
# ===========================================================
class_labels = ['Tay phải', 'Tay trái', 'Chân phải', 'Chân trái']

# Channel names & sampling rate
channel_names = epochs.info['ch_names']
fs = int(epochs.info['sfreq'])
n_channels = len(channel_names)


# ===========================================================
# 2️⃣ Hàm MVAR Yule–Walker
# ===========================================================
def estimate_mvar_yw(X, order):
    n_channels, n_samples = X.shape
    R = np.zeros((order + 1, n_channels, n_channels))

    for k in range(order + 1):
        denominator = (n_samples - k)
        if denominator <= 0:
            raise ValueError("Not enough samples for MVAR")
        R[k] = (X[:, k:] @ X[:, :n_samples - k].T) / denominator

    # Build block Toeplitz matrix
    R0 = np.hstack([R[i] for i in range(1, order + 1)])
    T = np.vstack([np.hstack([R[abs(i - j)] for j in range(1, order + 1)])
                   for i in range(1, order + 1)])

    A_vec = solve(T, R0.T).T
    A_matrices = A_vec.reshape(order, n_channels, n_channels)

    E = R[0] - sum(A_matrices[p] @ R[p + 1] for p in range(order))
    return A_matrices, E


# ===========================================================
# 3️⃣ Hàm tính dTF
# ===========================================================
def compute_dtf(A_matrices, fs=128, n_freqs=200):
    P, n, _ = A_matrices.shape
    freqs = np.linspace(0, fs / 2, n_freqs)
    dtf = np.zeros((n, n, n_freqs), float)
    I = np.eye(n, dtype=complex)

    for fi, f in enumerate(freqs):
        Af = I.copy()
        for p in range(P):
            Af -= A_matrices[p] * np.exp(-2j * np.pi * f * (p + 1) / fs)

        try:
            Hf = np.linalg.inv(Af)
        except np.linalg.LinAlgError:
            Hf = np.linalg.inv(Af + 1e-8 * np.eye(n))

        H2 = np.abs(Hf)**2
        denom = np.sum(H2, axis=0, keepdims=True)
        denom[denom == 0] = 1e-16
        dtf[:, :, fi] = H2 / denom

    return freqs, dtf


# ===========================================================
# 4️⃣ Hàm xử lý 1 class → output weighted matrix
# ===========================================================
def process_class(epochs_class, class_name, order=5):
    print(f"\n===== PROCESSING CLASS: {class_name} =====")

    # Lấy dữ liệu của class
    data = epochs_class.get_data()   # shape: (n_epochs, n_channels, n_times)

    # Ghép tất cả epochs → 1 chuỗi lớn
    eeg = data.reshape(-1, data.shape[1]).T

    print("Merged EEG shape:", eeg.shape)

    # --- MVAR ---
    A, E = estimate_mvar_yw(eeg, order=order)

    # --- dTF ---
    freqs, dtf = compute_dtf(A, fs=fs)

    # --- Alpha band ---
    alpha_idx = np.where((freqs >= 8) & (freqs <= 13))[0]
    weighted_matrix = np.mean(dtf[:, :, alpha_idx], axis=2)
    np.fill_diagonal(weighted_matrix, 0)

    # --- Save CSV ---
    csv_name = f"dTF_alpha_{class_name.replace(' ','_')}.csv"
    pd.DataFrame(weighted_matrix, index=channel_names, columns=channel_names)\
        .to_csv(csv_name, float_format="%.6f")

    print("Saved:", csv_name)

    # --- Plot ---
    plt.figure(figsize=(9, 7))
    plt.imshow(weighted_matrix, cmap='viridis', origin='lower')
    plt.colorbar(label='dTF (alpha)')
    plt.xticks(np.arange(n_channels), channel_names, rotation=90)
    plt.yticks(np.arange(n_channels), channel_names)
    plt.title(f"Connectivity dTF Alpha (8–13 Hz) – {class_name}")
    plt.tight_layout()
    plt.show()

    return weighted_matrix


# ===========================================================
# 5️⃣ Chạy cho cả 4 class
# ===========================================================
results = {}

for label in class_labels:
    epochs_class = epochs[label]
    results[label] = process_class(epochs_class, label)

print("\nHoàn tất: đã xuất 4 ma trận connectivity alpha cho 4 lớp.")
```

```python

import networkx as nx

# ===========================================================
# 6️⃣ Hàm tạo binary matrix theo threshold 5%
# ===========================================================

def apply_threshold(weighted_matrix, top_percent=0.05):
    """
    Giữ lại top X% connections mạnh nhất → binary matrix
    Diagonal đã = 0 nên không ảnh hưởng
    """
    mat = weighted_matrix.copy()
    # Lấy tất cả giá trị off-diagonal
    flat = mat[~np.eye(mat.shape[0], dtype=bool)]
    threshold_val = np.percentile(flat, (1 - top_percent) * 100)
    binary = (mat >= threshold_val).astype(int)
    np.fill_diagonal(binary, 0)
    return binary, threshold_val



# ===========================================================
# 7️⃣ Hàm tính Graph Theory metrics theo networkx
# ===========================================================
def compute_graph_metrics(binary_matrix, channel_names):
    G = nx.from_numpy_array(binary_matrix, create_using=nx.DiGraph)
    mapping = {i: channel_names[i] for i in range(len(channel_names))}
    G = nx.relabel_nodes(G, mapping)

    metrics = {
        "Degree"            : dict(G.degree()),
        "In-Degree"         : dict(G.in_degree()),
        "Out-Degree"        : dict(G.out_degree()),
        "Clustering"        : nx.clustering(G.to_undirected()),
        "Betweenness"       : nx.betweenness_centrality(G, normalized=True),
        "Global_Efficiency" : nx.global_efficiency(G.to_undirected()),
    }
    return G, metrics


# ===========================================================
# 8️⃣ Hàm plot metrics (bar charts)
# ===========================================================
def plot_metrics(metrics, channel_names, class_name, output_dir=""):
    fig, axes = plt.subplots(3, 1, figsize=(13, 11))
    fig.suptitle(f"Graph Metrics – {class_name} (Top 20% threshold)", fontsize=13)

    bar_items = [
        ("Degree",       "Degree",                  "steelblue"),
        ("Clustering",   "Clustering Coefficient",  "darkorange"),
        ("Betweenness",  "Betweenness Centrality",  "seagreen"),
    ]

    for ax, (key, ylabel, color) in zip(axes, bar_items):
        values = [metrics[key][ch] for ch in channel_names]
        ax.bar(channel_names, values, color=color)
        ax.set_ylabel(ylabel)
        ax.set_xticks(range(len(channel_names)))
        ax.set_xticklabels(channel_names, rotation=90)

    plt.tight_layout()
    fig_name = f"{output_dir}graph_metrics_{class_name.replace(' ', '_')}.png"
    plt.savefig(fig_name, dpi=150)
    plt.show()
    print(f"  Saved figure: {fig_name}")


# ===========================================================
# 9️⃣ Hàm vẽ network graph
# ===========================================================
def plot_network(G, metrics, class_name, output_dir=""):
    fig, ax = plt.subplots(figsize=(10, 8))   # ← khai báo ax rõ ràng

    pos = nx.circular_layout(G)

    node_size  = [metrics["Degree"][n] * 80 + 100 for n in G.nodes()]
    node_color = [metrics["Betweenness"][n] for n in G.nodes()]

    nx.draw_networkx_nodes(G, pos,
                           ax=ax,
                           node_size=node_size,
                           node_color=node_color,
                           cmap=plt.cm.YlOrRd,
                           alpha=0.9)
    nx.draw_networkx_labels(G, pos, ax=ax, font_size=7)
    nx.draw_networkx_edges(G, pos,
                           ax=ax,
                           edge_color='gray',
                           arrows=True,
                           arrowsize=10,
                           alpha=0.5,
                           width=0.8)

    sm = plt.cm.ScalarMappable(cmap=plt.cm.YlOrRd,
                                norm=plt.Normalize(vmin=min(node_color),
                                                   vmax=max(node_color)))
    fig.colorbar(sm, ax=ax, label="Betweenness Centrality")   # ← dùng fig.colorbar + ax=ax

    ax.set_title(f"Brain Network – {class_name} (Top 20%)")
    ax.axis('off')
    plt.tight_layout()

    fig_name = f"{output_dir}network_{class_name.replace(' ', '_')}.png"
    plt.savefig(fig_name, dpi=150)
    plt.show()
    print(f"  Saved figure: {fig_name}")



# ===========================================================
# 🔟 Hàm lưu CSV metrics
# ===========================================================
def save_metrics_csv(metrics, channel_names, class_name, output_dir=""):
    out_df = pd.DataFrame({
        "Channel"    : channel_names,
        "Degree"     : [metrics["Degree"][ch]       for ch in channel_names],
        "In_Degree"  : [metrics["In-Degree"][ch]    for ch in channel_names],
        "Out_Degree" : [metrics["Out-Degree"][ch]   for ch in channel_names],
        "Clustering" : [metrics["Clustering"][ch]   for ch in channel_names],
        "Betweenness": [metrics["Betweenness"][ch]  for ch in channel_names],
    })
    csv_name = f"{output_dir}graph_metrics_{class_name.replace(' ', '_')}.csv"
    out_df.to_csv(csv_name, index=False)
    print(f"  Saved CSV   : {csv_name}")
    print(f"  Global Efficiency = {metrics['Global_Efficiency']:.4f}")
    return out_df


# ===========================================================
# 1️⃣1️⃣ Chạy pipeline graph cho cả 4 class
# ===========================================================
graph_results = {}

for label in class_labels:
    print(f"\n========== GRAPH: {label} ==========")

    # Lấy weighted matrix từ kết quả MVAR+dTF trước đó
    weighted_matrix = results[label]

    # Threshold 20%
    binary_matrix, thr_val = apply_threshold(weighted_matrix, top_percent=0.20)
    print(f"  Threshold value: {thr_val:.6f} | Edges kept: {binary_matrix.sum()}")

    # Lưu binary matrix
    bin_csv = f"{output_dir}dTF_alpha_{label.replace(' ', '_')}_binary_top20.csv"
    pd.DataFrame(binary_matrix, index=channel_names, columns=channel_names) \
      .to_csv(bin_csv)
    print(f"  Saved binary: {bin_csv}")

    # Graph metrics
    G, metrics = compute_graph_metrics(binary_matrix, channel_names)

    # Plot bar charts
    plot_metrics(metrics, channel_names, label, output_dir)

    # Plot network
    plot_network(G, metrics, label, output_dir)

    # Save CSV
    df_metrics = save_metrics_csv(metrics, channel_names, label, output_dir)

    graph_results[label] = {
        "G"       : G,
        "metrics" : metrics,
        "binary"  : binary_matrix,
        "df"      : df_metrics,
    }

print("\n✅ Hoàn tất pipeline: MVAR → dTF → Graph Theory cho 4 lớp.")
```

```python
import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from scipy.linalg import solve
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

# ===========================================================
# Dùng lại các hàm đã có từ pipeline trước:
# estimate_mvar_yw, compute_dtf, apply_threshold
# ===========================================================

def estimate_mvar_yw_single(X, order):
    """MVAR cho 1 epoch đơn lẻ: X shape (n_channels, n_times)"""
    n_channels, n_samples = X.shape
    X = X - X.mean(axis=1, keepdims=True)   # demean
    R = np.zeros((order + 1, n_channels, n_channels))

    for k in range(order + 1):
        denom = n_samples - k
        if denom <= 0:
            raise ValueError("Epoch quá ngắn")
        R[k] = (X[:, k:] @ X[:, :n_samples - k].T) / denom

    R0 = np.hstack([R[i] for i in range(1, order + 1)])
    T  = np.vstack([np.hstack([R[abs(i - j)] for j in range(1, order + 1)])
                    for i in range(1, order + 1)])

    A_vec      = solve(T, R0.T).T
    A_matrices = A_vec.reshape(order, n_channels, n_channels)
    return A_matrices


# ===========================================================
# 1️⃣ Trích xuất features từ 1 epoch
# ===========================================================
def extract_features_single_epoch(X, order=5, fs=128, top_percent=0.20):
    """
    X     : (n_channels, n_times) - 1 epoch
    return: feature vector 1D
    """
    try:
        A = estimate_mvar_yw_single(X, order)
    except Exception:
        return None

    freqs, dtf = compute_dtf(A, fs=fs)

    # Alpha band
    alpha_idx = np.where((freqs >= 8) & (freqs <= 13))[0]
    W = np.mean(dtf[:, :, alpha_idx], axis=2)
    np.fill_diagonal(W, 0)

    # Binary threshold
    binary, _ = apply_threshold(W, top_percent=top_percent)

    # Graph
    G = nx.from_numpy_array(binary, create_using=nx.DiGraph)
    G_und = G.to_undirected()

    degree      = np.array([d for _, d in G.degree()])
    in_degree   = np.array([d for _, d in G.in_degree()])
    out_degree  = np.array([d for _, d in G.out_degree()])
    clustering  = np.array(list(nx.clustering(G_und).values()))
    betweenness = np.array(list(nx.betweenness_centrality(G, normalized=True).values()))
    global_eff  = np.array([nx.global_efficiency(G_und)])

    # Ghép thành 1 vector
    feature_vec = np.concatenate([
        degree,
        in_degree,
        out_degree,
        clustering,
        betweenness,
        global_eff
    ])
    return feature_vec


# ===========================================================
# 2️⃣ Build dataset X, y từ tất cả epochs
# ===========================================================
class_labels = ['Tay phải', 'Tay trái', 'Chân phải', 'Chân trái']
fs     = int(epochs.info['sfreq'])
order  = 5

X_list, y_list = [], []

for label in class_labels:
    epochs_class = epochs[label]
    data = epochs_class.get_data()   # (n_epochs, n_channels, n_times)
    print(f"\nProcessing [{label}]: {data.shape[0]} epochs")

    for i in range(data.shape[0]):
        feat = extract_features_single_epoch(
            data[i], order=order, fs=fs, top_percent=0.20
        )
        if feat is not None:
            X_list.append(feat)
            y_list.append(label)
        else:
            print(f"  ⚠️  Epoch {i} bị lỗi, bỏ qua")

X = np.array(X_list)
y = np.array(y_list)

print(f"\nFeature matrix X: {X.shape}")
print(f"Labels y        : {np.unique(y, return_counts=True)}")

# Tên features (để phân tích sau nếu cần)
n_ch = len(epochs.info['ch_names'])
ch   = epochs.info['ch_names']
feature_names = (
    [f"degree_{c}"      for c in ch] +
    [f"in_degree_{c}"   for c in ch] +
    [f"out_degree_{c}"  for c in ch] +
    [f"clustering_{c}"  for c in ch] +
    [f"betweenness_{c}" for c in ch] +
    ["global_efficiency"]
)
print(f"Số features: {len(feature_names)}")


# ===========================================================
# 3️⃣ SVM với Stratified K-Fold Cross-Validation
# ===========================================================
le = LabelEncoder()
y_enc = le.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

svm = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_results = cross_validate(
    svm, X_scaled, y_enc, cv=cv,
    scoring=['accuracy', 'f1_macro'],
    return_train_score=True
)

print("\n===== SVM Cross-Validation (5-Fold) =====")
print(f"Train Accuracy : {cv_results['train_accuracy'].mean():.4f} ± {cv_results['train_accuracy'].std():.4f}")
print(f"Test  Accuracy : {cv_results['test_accuracy'].mean():.4f}  ± {cv_results['test_accuracy'].std():.4f}")
print(f"Test  F1 Macro : {cv_results['test_f1_macro'].mean():.4f}  ± {cv_results['test_f1_macro'].std():.4f}")


# ===========================================================
# 4️⃣ Train toàn bộ data → Confusion Matrix
# ===========================================================
from sklearn.model_selection import cross_val_predict

y_pred = cross_val_predict(svm, X_scaled, y_enc, cv=cv)

print("\n===== Classification Report =====")
print(classification_report(y_enc, y_pred,
                             target_names=le.classes_))

# Confusion Matrix
cm = confusion_matrix(y_enc, y_pred)
fig, ax = plt.subplots(figsize=(7, 6))
disp = ConfusionMatrixDisplay(confusion_matrix=cm,
                               display_labels=le.classes_)
disp.plot(ax=ax, cmap='Blues', colorbar=False)
plt.title("Confusion Matrix – SVM (Graph Metrics Features)")
plt.tight_layout()
plt.savefig(f"{output_dir}confusion_matrix_SVM.png", dpi=150)
plt.show()


# ===========================================================
# 5️⃣ Lưu features ra CSV (tuỳ chọn)
# ===========================================================
feat_df = pd.DataFrame(X, columns=feature_names)
feat_df.insert(0, 'Label', y)
feat_df.to_csv(f"{output_dir}graph_features_all_epochs.csv", index=False)
print(f"\n✅ Saved features: {output_dir}graph_features_all_epochs.csv")
```
